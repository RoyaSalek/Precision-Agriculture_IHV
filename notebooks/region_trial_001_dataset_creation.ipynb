# Region Trial 001

## Purpose

Investigate how vineyard anomaly **regions** should be defined for AI model training.

This experiment uses images generated from the 470 Crew Leaf Extraction workflow as a starting dataset. The goal is not to build a production model, but to develop an understanding of:

* What constitutes a region
* How regions should be labeled
* Whether YOLO can learn meaningful anomaly patterns
* What dataset structure is required for future training efforts

## Data Source

* Source: 470 Crew Leaf Extraction experiment
* Status: Raw imagery
* Provenance (where the data came from): Exact source video currently unknown
* Storage: `data/raw/470_crew_leaf_extraction_frames_unknown_source/`

## Experiment Goals

1. Review extracted imagery
2. Define candidate anomaly regions
3. Create a small annotated dataset
4. Train a baseline YOLO model
5. Evaluate detections and region quality
6. Document findings and recommendations

## Expected Outputs

* Region definition proposal
* Annotated training dataset
* Baseline YOLO model
* Example detections
* Recommendations for future dataset creation
